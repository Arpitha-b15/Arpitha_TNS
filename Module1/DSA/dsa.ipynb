```python
#LIST

# Problem Statement:
# A food delivery company maintains a list of available delivery agents.
# New agents may join the company and existing agents may leave.
# Identify the suitable data type and implement it.
#
# Data Type Used: List
# Reason: List is mutable and allows adding and removing elements.

delivery_agents = ["Ravi", "Anu", "Kiran", "Priya"]

print("Available Agents:", delivery_agents)

delivery_agents.append("Arjun")

print("\nAfter New Agent Joins:")
print(delivery_agents)

delivery_agents.remove("Kiran")

print("\nAfter Agent Leaves:")
print(delivery_agents)

print("\nTotal Agents:", len(delivery_agents))
```

```python
#TUPLE

# Problem Statement:
# A satellite mission has fixed target planets approved before launch.
# These destinations cannot be modified after approval.
# Identify the suitable data type and implement it.
#
# Data Type Used: Tuple
# Reason: Tuple is immutable and protects fixed data.

target_planets = ("Mars", "Jupiter", "Saturn")

print("Mission Destinations:")

for planet in target_planets:
    print(planet)

print("\nTotal Destinations:", len(target_planets))
```

```python
#SET

# Problem Statement:
# An online coding contest stores participant usernames.
# A participant may register multiple times, but duplicate usernames
# should not be stored.
# Identify the suitable data type and implement it.
#
# Data Type Used: Set
# Reason: Set automatically removes duplicate values.

participants = {
    "arpitha",
    "rahul",
    "anu",
    "arpitha",
    "kiran"
}

print("Registered Participants:")
print(participants)

print("\nTotal Unique Participants:", len(participants))
```

```python
# DICTIONARY

# Problem Statement:
# A hospital stores patient IDs and corresponding patient names.
# The receptionist should quickly retrieve patient information
# using the patient ID.
# Identify the suitable data type and implement it.
#
# Data Type Used: Dictionary
# Reason: Dictionary stores data in key-value pairs for fast lookup.

patients = {
    101: "Arpitha",
    102: "Roopa",
    103: "Bharath"
}

print("Patient Records:")
print(patients)

patient_id = 102

print("\nSearching Patient ID:", patient_id)
print("Patient Name:", patients[patient_id])
```
